#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
library(tidymodels)
# Remove missing values
model_data <- march_tidy |>drop_na()
# Split data into training and testing sets
set.seed(1234)
march_split <- initial_split(model_data,prop = 0.75,strata = won_game)
march_training <- training(march_split)
march_testing <- testing(march_split)
# Create a recipe and scale the predictors
march_recipe <- recipe(won_game ~ seed + reg_wins_pct,data = march_training) |>step_normalize(all_predictors())
# Create a KNN classification model
knn_spec <- nearest_neighbor(weight_func = "rectangular",neighbors = tune()) |>set_engine("kknn") |>set_mode("classification")
# Create 5-fold cross-validation
set.seed(1234)
march_vfold <- vfold_cv(march_training,v = 5,strata = won_game)
# Create workflow
march_workflow <- workflow() |>add_recipe(march_recipe) |>add_model(knn_spec)
# Test different K values
set.seed(1234)
march_results <- march_workflow |>tune_grid(resamples = march_vfold,grid = tibble(neighbors = seq(1, 31, by = 2)),metrics = metric_set(accuracy))
# Find the best K value
best_k <- march_results |>select_best(metric = "accuracy")
best_k
# Build the final model
march_final_workflow <- march_workflow |>finalize_workflow(best_k)
# Train the model
march_final_fit <- march_final_workflow |>fit(data = march_training)
# Make predictions using testing data
march_predictions <- march_final_fit |>predict(march_testing) |>bind_cols(march_testing)
# Check model accuracy
march_accuracy <- march_predictions |>metrics(truth = won_game,estimate = .pred_class)
march_accuracy
# Create a confusion matrix
march_confusion <- march_predictions |>conf_mat(truth = won_game,estimate = .pred_class)
march_confusion
```
